```python

from __future__ import print_function, division
from collections import defaultdict, OrderedDict
import gzip
import pickle
import json
import os
import uproot
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
from coffea import hist, processor 
from coffea.hist import plot
from coffea.util import load, save
```

```python
hists = load('lep_monotop_v1_2017.scaled')
data={
    '2017':{
        'sre': 'SingleElectron',
        'srm': 'SingleMuon',
        'wjetm': 'SingleMuon',
        'ttbarm': 'SingleMuon',
        'wjete': 'SingleElectron',
        'ttbare': 'SingleElectron',
#         'zmcr': 'MET',
#         'zecr': 'EGamma',
#         'gcr': 'EGamma'
    }
}

data_hists   = hists['data']
bkg_hists    = hists['bkg']
signal_hists = hists['sig']
```

```python
proc_list = []
for i in bkg_hists['template'].integrate('region', 'wjete').integrate('systematic', 'nominal').integrate('mT')[()].values().keys():
        p = str(i[0])
        proc_list.append(p)
#         print(sum(bkg_hists['template_mt_dphi_met_lj'].integrate('region', 'wjete').integrate('systematic', 'nominal').integrate('process', p).integrate('mT').values()[()]))
proc_list
```

```python
from matplotlib.pyplot import figure
for var in bkg_hists.keys(): 
    if 'template' in var: continue
    if 'sumw' in var: continue
#     for r in ['wjete', 'wjetm', 'ttbare', 'ttbarm']:
    r='wjete'
    for proc in proc_list:
        fig, ax = plt.subplots(1, 1, figsize=(15,10),sharex=True)
        fig.subplots_adjust(hspace=.07)
        ax.set_yscale('log')
        try:
            plot.plot1d(bkg_hists[var].integrate('region', r)[proc],
                       fill_opts={'alpha': .5, 'edgecolor': (0,0,0,0.3)})
            plt.title('region: {} process: {}'.format(r,proc))
        except:
            print(proc)
```
