```python
def plotratio2d(numerator, denominator):
    NumeratorAxes = numerator.axes()
    DenominatorAxes = denominator.axes()
    
    # integer number of bins in this axis #
    NumeratorAxis1_BinNumber = NumeratorAxes[0].size - 3 # Subtract 3 to remove overflow
    NumeratorAxis2_BinNumber = NumeratorAxes[1].size - 3
    
    DenominatorAxis1_BinNumber = DenominatorAxes[0].size - 3 
    DenominatorAxis2_BinNumber = DenominatorAxes[1].size - 3 
    
    if(NumeratorAxis1_BinNumber != DenominatorAxis1_BinNumber 
       or NumeratorAxis2_BinNumber != DenominatorAxis2_BinNumber):
        raise Exception('Numerator and Denominator axes are different sizes; Cannot perform division.')
    else:
        Numerator = numerator.to_hist()
        Denominator = denominator.to_hist()

        ratio = Numerator / Denominator.values()
#         ratio_norm = ratio/max(ratio.flatten())
#         err_up, err_down = ratio_uncertainty(Numerator.values(), Denominator.values(), 'poisson-ratio')

        labels = []
        for ra in (ratio.values().ravel()):
            ra = f'{ra:.2f}'
            st = '$'+ra+'$'
            labels.append(st)
        labels = np.array(labels).reshape(NumeratorAxis1_BinNumber,NumeratorAxis2_BinNumber)
        
    return(ratio, labels)
```

```python
import mplhep as hep
from hist.intervals import ratio_uncertainty

mt_bins = np.linspace(0,600,11)
dphi_met_lj_bins = np.linspace(0,3.5,16)
dphi_lep_etmiss_bins = np.linspace(0, 3.5, 16)
et_miss_bins = [0,60, 120, 180, 240, 300, 360, 420, 480, 540, 600]

bkg_hists['template_etmiss_dphi_lep_etmiss']=bkg_hists['template_etmiss_dphi_lep_etmiss'].rebin("eT_miss", hist.Bin("et_miss_new", "rebinned et_miss value", et_miss_bins))
bkg_hists['template_etmiss_dphi_lep_etmiss']=bkg_hists['template_etmiss_dphi_lep_etmiss'].rebin("dphi_lep_etmiss", hist.Bin("dphi_lep_etmiss_new", "rebinned dphi_lep_etmiss value", dphi_lep_etmiss_bins))
data_hists['template_etmiss_dphi_lep_etmiss']=data_hists['template_etmiss_dphi_lep_etmiss'].rebin("eT_miss", hist.Bin("et_miss_new", "rebinned et_miss value", et_miss_bins))
data_hists['template_etmiss_dphi_lep_etmiss']=data_hists['template_etmiss_dphi_lep_etmiss'].rebin("dphi_lep_etmiss", hist.Bin("dphi_lep_etmiss_new", "rebinned dphi_lep_etmiss value", dphi_lep_etmiss_bins))        


bkg_hists['template_mt_dphi_met_lj']=bkg_hists['template_mt_dphi_met_lj'].rebin("mT", hist.Bin("mT_new", "rebinned mt value", mt_bins))
bkg_hists['template_mt_dphi_met_lj']=bkg_hists['template_mt_dphi_met_lj'].rebin("dphi_Met_LJ", hist.Bin("dphi_Met_LJ_new", "rebinned dphi_Met_LJ value", dphi_met_lj_bins))
data_hists['template_mt_dphi_met_lj']=data_hists['template_mt_dphi_met_lj'].rebin("mT", hist.Bin("mT_new", "rebinned mt value", mt_bins))
data_hists['template_mt_dphi_met_lj']=data_hists['template_mt_dphi_met_lj'].rebin("dphi_Met_LJ", hist.Bin("dphi_Met_LJ_new", "rebinned ddphi_Met_LJ value", dphi_met_lj_bins))        

bkg_hists['template_mt_dphi_lep_etmiss']=bkg_hists['template_mt_dphi_lep_etmiss'].rebin("mT", hist.Bin("mT_new", "rebinned mt value", mt_bins))
bkg_hists['template_mt_dphi_lep_etmiss']=bkg_hists['template_mt_dphi_lep_etmiss'].rebin("dphi_lep_etmiss", hist.Bin("dphi_lep_etmiss_new", "rebinned dphi_Met_LJ value", dphi_lep_etmiss_bins))
data_hists['template_mt_dphi_lep_etmiss']=data_hists['template_mt_dphi_lep_etmiss'].rebin("mT", hist.Bin("mT_new", "rebinned mt value", mt_bins))
data_hists['template_mt_dphi_lep_etmiss']=data_hists['template_mt_dphi_lep_etmiss'].rebin("dphi_lep_etmiss", hist.Bin("dphi_lep_etmiss_new", "rebinned ddphi_Met_LJ value", dphi_lep_etmiss_bins))        


for hist in ['template_mt_dphi_met_lj', 'template_etmiss_dphi_lep_etmiss', 'template_mt_dphi_lep_etmiss']:
    for reg in ['wjete', 'wjetm', 'ttbare', 'ttbarm']:

        
        
        if 'e' in reg:
            a = data_hists[hist].integrate('region', reg).integrate('systematic', 'nominal')['SingleElectron'].integrate('process')
        if 'm' in reg:
            a = data_hists[hist].integrate('region', reg).integrate('systematic', 'nominal')['SingleMuon'].integrate('process')
#         a = data_hists[hist].integrate('process').integrate('region', reg).integrate('systematic', 'nominal')
        b = bkg_hists[hist].integrate('process').integrate('region', reg).integrate('systematic', 'nominal')
#         ratio = a / b.values()
        # err_up, err_down = ratio_uncertainty(a.values(), b.values(), 'poisson-ratio')
        ratio, labels = plotratio2d(a,b)
        fig, ax = plt.subplots(1, 1, figsize=(15,10),sharex=True)
        fig.subplots_adjust(hspace=.07)
#         print(np.nanmax(ratio.values().flatten()))
        arr =  ratio.values().flatten()
        arr[~np.isfinite(arr)] = 0
        norm = max(arr)
#         print(np.round(norm,2))
        hep.hist2dplot(ratio, labels=labels, cmap='viridis')
        title = 'Data/MC ratio ' + reg 
#         print(reg)
        plt.title(title,fontdict = {'fontsize' : 10})
```
