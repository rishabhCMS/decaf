# 2018UL

```python
hists = load('../../../hists/UL_lep_monotop_v1_test_2018.scaled')
data={
    '2018':{
        'sre': 'EGamma',
        'srm': 'SingleMuon',
        'wjetm': 'SingleMuon',
        'ttbarm': 'SingleMuon',
        'wjete': 'EGamma',
        'ttbare': 'EGamma',
#         'zmcr': 'MET',
#         'zecr': 'EGamma',
#         'gcr': 'EGamma'
    }
}

data_hists   = hists['data']
bkg_hists    = hists['bkg']
signal_hists = hists['sig']
```

```python
bkg_hists
```

```python
sys = 'nominal'
region = 'wjete'
for region in ['wjete', 'wjetm', 'ttbare', 'ttbarm']:
    print('************************')
    print("******",region, '*******')
    if region[-1]=='e':
        data='EGamma'
    if region[-1]=='m':
        data='SingleMuon'
    SUM_data = sum(data_hists['mT'].integrate('systematic', sys).integrate('region', region).integrate('process', data).values()[()])
#     print(data,':', SUM_data)

    tmp = 0
    for process in bkg_hists['mT'].identifiers('process'):

        SUM_bkg = sum(bkg_hists['mT'].integrate('systematic', sys).integrate('region', region).integrate('process', process).values()[()])
        tmp+=SUM_bkg
        print(process,":",SUM_bkg)
    print("Sum MC:", tmp)
    print(data,':', SUM_data)
    print("Data/MC:", SUM_data/tmp)
    print('#########################')
```
